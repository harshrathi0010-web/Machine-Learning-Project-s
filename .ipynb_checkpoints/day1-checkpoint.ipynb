# Types of machine learning
1. Supervised learning (Labelled dataset ( Input+Output)
    -  Regression (Predict contigous value ( House Price Prediction,Stock Price Prediction )
    -  Classification( categorize/classify ) Yes/NO, True/False,0/1, Spam/Not spam

2.  Unsupervised learning (Unlabelled datset (Input))
-   Dimensional Reduction(Reduce the number of feature)
    -   Feature Selection( Remove the unwanteald columns,use less feature)
    -   Feature Extraction ( Merge Features--Algo,PCA,UMAP etc )
    -   Clustering( Grouping -- K means , DBscan)
    -   Assocation rules -Relationship between two data points ,items sets( Recommendation models)
    -    
5. Reinforcement learning (without dataset)
    - Trial and error( Reward and Punishment)
    - Video Games
