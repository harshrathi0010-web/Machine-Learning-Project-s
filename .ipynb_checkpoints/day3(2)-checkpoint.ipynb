# Ordinal encoding

```python
newdf.Segment.unique()
```

```python
dic={"Government":4,"Midmarket":2,"Channel Partners":3,"Enterprise":0,"Small Business":1}
```

```python
newdf["Segment"]=newdf["Segment"].map(dic)
```

```python
newdf.head(20)
```

```python
newdf.Country.unique()
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
oe=OrdinalEncoder(categories=["United States of America","Germany","France","Mexico",'Canada'])
```

```python
newdf['Country']=newdf['Country']=oe.fit_transform(newdf[['Country']])
```
